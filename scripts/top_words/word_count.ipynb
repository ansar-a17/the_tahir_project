```python
from pathlib import Path
import json
import stanza

stanza.download("ur")
nlp = stanza.Pipeline(
    "ur",
    processors="tokenize,pos",
    use_gpu=True
)
```

```python
files = [
    f for f in Path("transcripts").glob("*.txt")
    if "segments" not in f.name
]

len(files)
```

```python
from collections import Counter

words = Counter()
count = 1

for file in files:

    print(f"{file.stem} {count/len(files)*100:.3f}")

    text = file.read_text(encoding="utf-8")

    doc = nlp(text)

    for sent in doc.sentences:
        for token in sent.words:
            if token.upos in {"NOUN", "PROPN"}:
                words[token.text] += 1

    count += 1
```

```python
with open("word_counts.json", "w", encoding="utf-8") as f:
        json.dump(words, f, ensure_ascii=False, indent=4)
```

```python
top_20 = sorted(words.items(), key=lambda x: x[1], reverse=True)[:20]
top_20
```

```python
with open("top_20_words.json", "w", encoding="utf-8") as f:
        json.dump(top_20, f, ensure_ascii=False, indent=4)
```
